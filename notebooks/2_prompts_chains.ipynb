# Prompts template basically dynamic prompts

### basically we use dynamic prompts or u can say prompts template so that we dont have to call llms again and again for diifernt works such as translator or semior pyhton dev or somethig like that

```python
from langchain_google_genai import GoogleGenerativeAI 
from langchain_core.prompts import ChatPromptTemplate
```

```python
llm = GoogleGenerativeAI(model='gemini-2.5-flash-lite')
```

```python
prompts = ChatPromptTemplate.from_messages([
     {"role": "system" , "content" :  "You Are A translator and u translate input to {language} ."},
    {"role": "user", "content" : "{query} ?"},
])
# so basically  .from_messages is a class method of chat prompt template , and it is usefull for dynamic prompts and we will pass the same list of dictinories as we pass in static prompt 
```

```python
prompts
```

```python
final_prompts = prompts.format_messages(language = "hindi" , query = "i love to play football ")

final_prompts
```

```python
ress = llm.invoke(final_prompts)
```

```python
print(ress)
```

## what are runnables 

## Runnables in LangChain are a core concept that represent modular, composable units of work—any component that can take input, process it, and produce output.  They serve as the building blocks for creating complex AI workflows and chains, enabling seamless integration of various components like language models (LLMs), prompt templates, output parsers, retrievers, and custom functions. 

##
basic variables of langchains and its working


1. Prompt Variables → fill-in blanks
   {name}, {topic}

2. Chain Variables → input/output names
   input_keys, output_keys

3. LLM Variables → answer style
   temperature, max_tokens, top_p

4. Tool Variables → extra functions
   name, func, description

5. Agent Variables → tool chooser
   agent_type, tools, memory

6. Memory Variables → chat history
   memory_key, input_key, output_key

7. Retriever Variables → find documents
   k, search_type, embeddings

8. Document Variables → text info
   page_content, metadata

9. Callback Variables → progress updates
   logs, streaming



Prompt → blanks
Chain → mapping
LLM → style
Tool → helper
Agent → chooser
Memory → history
Retriever → finder
Document → data
Callback → updates


## What are Chains in LangChain?
 ## Chains in LangChain are a way to connect steps together so the output of one step becomes the input of the next.

## 👉 Think of a chain as a pipeline.

```python
import langchain
import langchain_core

print(langchain.__version__)
print(langchain_core.__version__)
```

```python
prompts =ChatPromptTemplate.from_messages([
    ("system", "you are a translator . You translate input to  {language} only translate donr say anything other than this ?"),
    ("user" , "{query}")
])
```

```python
chains = prompts | llm 
```

```python
res = chains.invoke({"language" : "hinglish","query":"hii how are you what are you doing man?"} )
```

```python
print(res)
```
