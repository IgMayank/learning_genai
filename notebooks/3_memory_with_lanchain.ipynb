## Manage history of llm so that it can remember things

```python
history = []
while True:
    query= input("user: ")
    print("User: ",query)
    if query.lower() in ['q','quit','exit',"escape","bye","tata"]:
        print("AI: Good Bye SIR!")
        break
    history.append(("user",query))
   
    res  = llm.invoke(history)
    history.append(("ai",res))

    print("AI: ",res, "\n")
```
