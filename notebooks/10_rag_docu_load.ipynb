### Text Loader 

```python
from langchain_community.document_loaders import TextLoader
```

### PDF loader

```python
from langchain_community.document_loaders import PyPDFLoader

loader_pdf = PyPDFLoader("../data/abc.pdf")

doc = loader_pdf.load()
two_doc=doc[0]
print(two_doc)
```

### CSV loader

```python
from langchain_community.document_loaders import CSVLoader
loader_csv = CSVLoader("../data/spam.csv")
csv = loader_csv.load()
print(csv)
```
