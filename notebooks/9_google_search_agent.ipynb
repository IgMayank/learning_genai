```python
from langchain_community.utilities import GoogleSerperAPIWrapper
from langchain_groq import ChatGroq
from langchain.agents import create_agent 
from langgraph.checkpoint.memory import MemorySaver
```

## for the memory part we imported MemorySaver from langgraph and created its object under name checkpointer in the agent function and also in the response section u have to initiliaze a thread (for the memory purpose ) if u change the thread id it will completly allocate to somewhere else and the agent wont have the context of previous thread or memory 

```python
llm = ChatGroq(model="openai/gpt-oss-20b")
```

```python
search = GoogleSerperAPIWrapper()
```

```python
agent = create_agent(
    model = llm,
    tools=[search.run],
    system_prompt="you are an agent who can search on google for any  question",
    checkpointer=MemorySaver()
)
```

```python
question = " current temprature of mukesh pg sector 67 mohali "
```

```python
response = agent.invoke(
    {"messages":[{"role":"user","content":question}]},
    {"configurable":{"thread_id":"anything"}},
                        )
```

```python
print(response["messages"][-1].content)
```
