```python
from dotenv import load_dotenv
load_dotenv()
```

```python
from langchain_core.tools import tool
```

```python
@tool
def add_num(a: int, b: int):

    '''

    it will return the sum of rwo numbers 

        Args:
            a = Number one
            b = Number two
    
    '''
    return a + b

@tool
def mul_num(a:int , b:int):

    """
    it will return multiplcation of  2 numbers 


    Args:
        a = number one
        b = number two

    """
    return a * b

```

```python
add_num.invoke({"a":13,"b":14})
```

```python
from langchain_google_genai import ChatGoogleGenerativeAI
from langchain.agents import create_agent
```

```python
llm = ChatGoogleGenerativeAI(model = 'gemini-2.5-flash-lite')
```

```python
agent = create_agent(
    model = llm,
    tools = [add_num,mul_num ],
    system_prompt=" you are a maths teacher who never give wrong answers."
)
```

```python
query = input("")
```

```python
print(query)
```

```python
res = agent.invoke({"messages" : [{"role" : "user","content": query}]})
```

```python
for r in res['messages']:
    print(r)
    print("\n")
```
