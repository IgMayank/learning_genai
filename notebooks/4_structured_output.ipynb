### Basically how we can an strucutured output from model instead of always getting an string as an output

### we have a specific lib in python known as pydantic that is used to get structured op lets understand more 

```python
from pydantic import Field ,BaseModel

class structured_output(BaseModel):
    Name:str  = Field(description="Complete Name")
    email:str = Field(description="Email address")
    age:int = Field(description="Age")



Structured_llm = llm.with_structured_output(structured_output)
```
