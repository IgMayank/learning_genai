# so as u can see we were expecting age as integer but it comes out as a string , so to fix this issue we use pydantic

```python
from pydantic import BaseModel
```

```python
class user(BaseModel):
    name : str 
    age : int 
```

```python
use = user(name="mayank",age =20)

```

# Field validation

```python
from pydantic import Field

class Product(BaseModel):

    title : str = Field(max_length=10)
    des : str = Field(min_length=20)
    price : int = Field(gt=0 , le=100)
```

```python
prouct= Product(title="samsung ",des="good phone under 100 ruppess",price=99)
prouct
```

```python
class order(BaseModel):
    id : str = Field(min_length=6,max_length=12)
    product : Product
```

```python
order = order(id="bfhjbh332" , product=prouct)
```

```python
order
```

```python
order.product.title
order.product.des
order.product.price
# u can use each line
```

# Typing Library

## if we want to give class as list we will use this librabry

```python
from typing import List
class carts(BaseModel):
    id : str = Field(min_length=6,max_length=12)
    items : List[Product]
    # we using list here cause we gonna give list of product as input and class can not be given as input of list without this library
    comment : str = Field(
        default="no comment"
    )
    #  we using default here cause comment is optional and user not always type in some comment , so for something that is optional we will use default 
```

```python
prouct_2= Product(title="iphone 18 ",des="good phone under 100 ruppess",price=99)
```

```python
cart = carts(id="fuydw7",items=[prouct,prouct_2],comment="its a good devce")
cart
```

```python
for product in cart.items:
    print(product.title)
```

## annotated module inside typing library

### we will use annotated module a lot of time in langgraph

```python
#
#   we can use annotated to build common validadtor for eg we wanna creae=te a validator that always store list of products as we do above  so how wull we do it
from typing import Annotated
common_validator = Annotated[list[Product],Field(default="")]
# after list[product] the next string in basically metadata that we define ourselves , we can also use field(default="") if we dont wanna give any metadata
# lets understand with same example we do above
```

```python

class carts(BaseModel):
    id : str = Field(min_length=6,max_length=12)
    items : common_validator
    # we using list here cause we gonna give list of product as input and class can not be given as input of list without this library
    comment : str = Field(
        default="no comment"
    )
    #  we using default here cause comment is optional and user not always type in some comment , so for something that is optional we will use default 
```

```python
class order(BaseModel):
    products : common_validator
```

```python
ord = order(products=[prouct,prouct_2])
```
