# Text Splitting

```python
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
text ="Massive Third-Party Ecosystem: Libraries for ML, \
    web dev, APIs, automation, data science, etc." 
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size =50, chunk_overlap = 20)
```

```python
chunk = splitter.split_text(text)
```

# Document  text loader 

```python
from langchain_community.document_loaders import TextLoader , PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
```

```python
loader = TextLoader("../data/python.txt")
texts = loader.load()
```

```python
pdf_loader = PyPDFLoader("../data/abc.pdf")
pdfss=pdf_loader.load()
```

```python
splitter = RecursiveCharacterTextSplitter(chunk_size = 500,chunk_overlap=100)
```

```python
text_chunks = splitter.split_documents(texts)
print(len(text_chunks))
```

```python
text_chunks
```

```python
pdf_chunks = splitter.split_documents(pdfss)
len(pdf_chunks)
```
